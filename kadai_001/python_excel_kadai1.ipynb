```python

# openpyxlライブラリからWorkbookを読み込む
from openpyxl import Workbook

# 文字の書式設定や配置、罫線に使用する機能を読み込む
from openpyxl.styles import Font, Alignment, Border, Side

# 現在の日付を取得する機能を読み込む
from datetime import date


# Excelブックとワークシートを作成する

# 新しいExcelブックを作成する
wb = Workbook()

# 操作するワークシートを取得する
ws = wb.active

# ワークシートの名前を「請求書」に変更する
ws.title = "請求書"


# 請求書の基本情報を入力する

# セルの位置と入力する値を辞書で管理する
basic_info = {
    "B2": "請求書",
    "B4": "株式会社ABC",
    "B5": "〒101-0022 東京都千代田区神田練塀町300",
    "B6": "TEL:03-1234-5678 FAX:03-1234-5678",
    "B7": "担当者名：鈴木　一郎様",
    "F4": "No.",
    "G4": "001",
    "F5": "日付",
    "B15": "小計",
    "B16": "消費税",
    "B17": "合計",
    "E13": 35000,
    "E15": 35000,
    "E16": 3500,
    "E17": 38500,
}

# 辞書に登録した情報をExcelのセルに入力する
for cell_address, value in basic_info.items():
    ws[cell_address] = value


# 現在の日付をG5に入力する

# date.today()で現在の日付を取得する
ws["G5"] = date.today()

# 日付を「西暦年/月/日」の形式で表示する
ws["G5"].number_format = "yyyy/mm/dd"


# 指定されたセルを結合する

# B5からE5までのセルを結合する
ws.merge_cells("B5:E5")

# B6からE6までのセルを結合する
ws.merge_cells("B6:E6")


# 商品一覧の見出しを入力する

# 商品一覧の見出しをリストで管理する
headers = ["商品名", "数量", "単価", "金額"]

# 見出しをB10からE10まで順番に入力する
for column_number, header in enumerate(headers, start=2):
    ws.cell(row=10, column=column_number, value=header)


# 商品データを入力する

# 商品情報をリストで管理する
products = [
    ["商品A", 2, 10000, 20000],
    ["商品B", 1, 15000, 15000],
]

# 商品データを11行目から順番に入力する
for row_number, product in enumerate(products, start=11):

    # 商品情報をB列からE列まで順番に入力する
    for column_number, value in enumerate(product, start=2):

        # 指定した行番号・列番号のセルに値を入力する
        ws.cell(
            row=row_number,
            column=column_number,
            value=value
        )


# 文字の大きさや配置、罫線を設定する

# B2の文字を18ポイント、太字にする
ws["B2"].font = Font(size=18, bold=True)

# 細い罫線を作成する
thin_side = Side(style="thin")

# 商品一覧のB10からE12までのセルに書式を設定する
for row in ws.iter_rows(
    min_row=10,
    max_row=12,
    min_col=2,
    max_col=5
):
    for cell in row:

        # 文字を中央揃え、上下中央に配置する
        cell.alignment = Alignment(
            horizontal="center",
            vertical="center"
        )

        # セルの上下左右に罫線を設定する
        cell.border = Border(
            left=thin_side,
            right=thin_side,
            top=thin_side,
            bottom=thin_side
        )


# 金額に3桁区切りのカンマを設定する

for cell_address in ["E13", "E15", "E16", "E17"]:
    ws[cell_address].number_format = "#,##0"


# Excelの列幅と行の高さを設定する

# 各列の幅を辞書で管理する
column_widths = {
    "B": 20,
    "C": 12,
    "D": 15,
    "E": 15,
    "F": 10,
    "G": 18,
}

# 辞書の情報を使って列幅を設定する
for column_letter, width in column_widths.items():
    ws.column_dimensions[column_letter].width = width

# 2行目の高さを30に設定する
ws.row_dimensions[2].height = 30


# Excelファイルを日付付きのファイル名で保存する

# 現在の日付を「年月日」の形式で取得する
today = date.today().strftime("%Y%m%d")

# 日付を含めたファイル名を作成する
file_name = f"請求書_{today}.xlsx"

# Excelファイルを保存する
wb.save(file_name)

# 処理が完了したことを表示する
print(f"{file_name}を作成しました。")
```
